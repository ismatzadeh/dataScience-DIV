```python
def elifbaYarat():
    import string
    orta = len(string.ascii_letters)//2
    kicikHerfler = string.ascii_letters[:orta]
    ## print(kicikHerfler)
    boyukHerfler = string.ascii_letters[orta:]
    ## print(boyukHerfler)

    boshDict = {}
    for ind in range(0, orta):
        boshDict[kicikHerfler[ind]] = boyukHerfler[ind]
    return {kicikHerfler[ind]:boyukHerfler[ind] for ind in range(orta)}

elifba=elifbaYarat()
elifba

dict(zip(kicikHerfler, boyukHerfler))
```
