```python
import pandas as pd
maghaza = {
    'kartof' : 0.8,
    'soghan' : 1.4,
    'alma' : 1.5,
    'banan': 2.7
}

maghaza.keys(), maghaza.values()
```

```python
mehsullar = pd.Series(maghaza.keys())
qiymatlar = pd.Series(maghaza.values())

dfMaghaza = pd.DataFrame({
    'mahsul': mehsullar,
    'qiymat': qiymatlar,
})

maskKartof = dfMaghaza['mahsul'] == 'kartof'
maskSoghan = dfMaghaza['mahsul'] == 'soghan'
dfMaghaza.loc[maskKartof | maskSoghan, 'qiymat']*3
```
