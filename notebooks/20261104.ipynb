#### 557. Reverse Words in a String III

```python
s = 'I will do my best at any time point.'

class Solution:
    def reverseWords(self, s: str) -> str:
        chars = list(s)
        n = len(chars)
        start = 0

        for i in range(n + 1):
            if i == n or chars[i] == " ":
                l, r = start, i - 1
                while l < r:
                    chars[l], chars[r] = chars[r], chars[l]
                    l += 1
                    r -= 1
                start = i + 1

        return "".join(chars)

sol = Solution()
solvedS = sol.reverseWords(s)
print(solvedS)
```
