```python
class Adam:
    def __init__(self, adamYashayir=True):
        self.yashayir = adamYashayir
        self.car = None
    def carAl(self, aldighiCar):
        self.car = aldighiCar
        aldighiCar.sahib = self

class Car:
    def __init__(self, aMarka=None, aModel=None, aVeziyyet=None, aKilometr=None, aSahibi=None):
        self.marka = aMarka
        self.model = aModel
        self.veziyyet = aVeziyyet
        self.kilometr = 0 if self.veziyyet == 'Salon' else aKilometr
        self.sahib = aSahibi 
    def sur(self, necheKilometr=1):
        if self.kilometr is not None:
            self.kilometr = self.kilometr + necheKilometr
    def vur(self, neyise=None):
        self.veziyyet = 'Qezali'
        self.sahib.yashayir = False
        if isinstance(neyise, Adam):
            neyise.yashayir = False
        elif isinstance(neyise, Car):
            neyise.veziyyet = 'Qezali'

adam1 = Adam()
mashin2 = Car(aMarka = 'Mercedes', aModel = 'E220', aKilometr = 391000 )
seninMashinin = Car(aMarka = 'Hyundai', aModel = 'Elantra')
mashin1 = Car(aVeziyyet = 'Salon')

mashin2.marka
mashin1.marka

mashin1.marka = 'Opel'
mashin1.model = 'Vectra'

mashin2.sur(10)
mashin2.kilometr
mashin1.sur(10)

print(adam1.yashayir)
print(mashin1.sahib)
adam1.carAl(mashin1)
mashin1.vur(adam1)
print(adam1.yashayir)
print


# leyla = Adam()
# leyla.carAl(mashin1)
# leyla.car
# shehriyar = Adam()
# print(leyla.yashayir)
```
