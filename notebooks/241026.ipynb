```python
# def fibonacci(r1, r2)

#     f = r1+r2
#     r1, r2 = r2, f
#     f = r1+r2

def nciFibonacciTap(necenci):
    r1 = 0
    r2 = 1
    n=2

    for i in range(necenci-2):
        f = r1+r2
        r1, r2 = r2, f
        n+=1
    return n, f

nciFibonacciTap(19)
```

```python

def nciFibonacciTap():
    r1 = 0
    r2 = 1
    n=2

    while True:
        f = r1+r2
        r1, r2 = r2, f
        n+=1
        yield n, f

fibonacciGen = nciFibonacciTap()
next(fibonacciGen)
```

```python
def to_superscript(number):
    superscript_map = str.maketrans("0123456789-", "⁰¹²³⁴⁵⁶⁷⁸⁹⁻")
    return str(number).translate(superscript_map)

to_superscript(1455)  # '¹²³'
```
