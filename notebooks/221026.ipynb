```python
class Circle:
    pi = 3.14

    def __init__(self, r):
        self.radius = r   
    @property
    def diameter(self):
        return self.radius*2 
    @staticmethod
    def calculate_area(r):
        return Circle.pi * (r**2) 
    #    self.area = Circle.pi * (r**2)

cl = Circle(r=3)
c2 = Circle(r=4)

Circle.calculate_area(r=3)
```

```python
class Square:
    def __init__(self, a):
        self.side = a

    @property
    def area(self):
        return (self.side)**2

    @staticmethod
    def calculate_area(a):
        return a**2

Square.calculate_area(a=12)

# square1 = Square(a=5)
# square1.area
# print(square1.area())
# square1.side = 10
# print(square1.area())
```

```python
class Cylinder:
    pi = 3.14    
    
    def __init__(self, r, h):
        
        self.radius = r
        self.height = h
    
    @property
    def volume(self):
        return self.pi * (self.radius**2) * self.height

    @staticmethod
    def calculateVolume(r, h):
        return Cylinder.pi * (r**2) * h

        

# slindr1 = Cylinder(r=2, h=5)
# slindr1.height
# slindr1.radius
# slindr1.volume

Cylinder.calculateVolume(r=3, h=10)
```

```python
class Cylinder:
    n = 0
    pi = 3.14    
    
    def __init__(self, r, h):
        
        self.radius = r
        self.height = h
        Cylinder.n = Cylinder.n + 1
    
    @property
    def volume(self):
        return self.pi * (self.radius**2) * self.height

    @staticmethod
    def calculateVolume(r, h):
        return Cylinder.pi * (r**2) * h
    @classmethod
    def resetn(cls):
        cls.n = 0
        return cls.n
        

# slindr1 = Cylinder(r=2, h=5)
# slindr1.height
# slindr1.radius
# slindr1.volume

slindr1 = Cylinder(r=2, h=5)
slindr2 = Cylinder(r=3, h=4)
slindr3 = Cylinder(r=1, h=10)
print(Cylinder.n)
Cylinder.resetn()
# print(Cylinder.n)
```

```python
class Rectangle:
    def __init__(self, a ,b):
        self.length = a
        self.width = b
    
    @property
    def area(self):
        return self.length*self.width

    @staticmethod
    def calculateArea(a, b):
        return a*b

    @classmethod
    def calculatePerimeter(cls, a, b):
        return 2*a + 2*b

# r1 = Rectangle(a=10, b=4)
# r1.width
# r1.length
# r1.area

Rectangle.calculateArea(a=5, b=39)
Rectangle.calculatePerimeter(a=5, b=39)
```
