#### 3. print() ilə return arasındakı fərq nədir?

#### 4. Aşağıdakı kod nə çap edəcək və niyə?

        def f(x):
            x * 2
        print(f(5))

```python
# Return yazilmadigi ucun deyer None olaraq qayidacaq. Cunki funksiyanin cixish deyerini geri qaytarmir. 

def f(x):
    x * 2
print(f(5))
```

#### 5. Default parametr nədir? def greet(name="Qonaq") çağırılanda nə baş verir?

```python
def greet(name='Qonaq'):
    return print(name)
greet()
```
